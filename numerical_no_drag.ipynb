```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
h = 0.01 #step
t = np.arange(0, 2 + h, h)
Y = np.zeros(len(t)) #create array (similar to Y[]={0} in C)
m = np.zeros(len(t))
m[0] = 1
Y[0] = 0 #intial value
ve=500
me=0.2
g=9.8
print(t)
```

```python
v_exact = ve * np.log(m[0] / (m[0] - me * t)) - g * t #exact solution from
for i in range(len(t) - 1):
    m[i+1] = m[i] - 0.2 * h
    Y[i + 1] = Y[i] + h * ((ve * me) / m[i] - g) #euler method
print(Y)
print(v_exact)
```

```python
plt.plot(t, Y, label='Euler method')
plt.plot(t, v_exact, label='Exact solution', linestyle='--', color='orange')

# max value
max_y = max(np.max(Y), np.max(v_exact))
ticks = [t for t in ticks if t < max_y]
ticks.append(max_y)
plt.yticks(ticks)

plt.title('Euler method vs Exact solution')
plt.xlabel('t (s)')
plt.ylabel('v (m/s)')
plt.grid(True)
plt.legend()
plt.show()
```
